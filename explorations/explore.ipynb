```python
from pydantic_ai import Agent
from pydantic import BaseModel
import os
from dotenv import load_dotenv

load_dotenv()

class SimpleResponse(BaseModel):
    answer: str
    confidence: str 

agent = Agent(
    model="google-gla:gemini-2.5-flash",
    system_prompt="You are a helpful assistant. Always rate your confidence in your answer.",
    output_type=SimpleResponse
)
```

```python
result = await agent.run("What is the capital of Sweden?")

print("Type of result:", type(result))
print("\n" + "="*50)
print("Result object attributes:")
print(dir(result))
```

```python
print("=== result.output ===")
print("Type:", type(result.output))
print("\nOutput:", result.output)
print("\n--- Accessing fields ---") 
print("Answer:", result.output.answer)
print("Confidence:", result.output.confidence)

print("\n--- As dict ---")
print(result.output.model_dump())
```
